# Importing Datasets with Pandas

## Introduction

In this article, we'll learn three ways to import datasets into Pandas:

1. **Importing CSV files**

## 1. Importing CSV Files

CSV (Comma-Separated Values) files are everywhere. They’re simple, lightweight, and probably the first data format you’ll encounter in the wild as a beginner.

For example, we found a `pokemon.csv` file containing information about Pokémon on Kaggle, an amazing resource for the Data Science community.

The beginning of the CSV file looks something like this:

```csv
pokedex_number,name,type1,type2,attack,defense,speed
1,Bulbasaur,grass,poison,49,49,45
2,Ivysaur,grass,poison,62,63,60
3,Venusaur,grass,poison,82,83,80
4,Charmander,fire,,52,43,65
5,Charmeleon,fire,,64,58,80
6,Charizard,fire,flying,84,78,100

```

Each line represents a row. Notice that the first row contains the column names; this is the **header**.

The following rows contain data separated by commas, hence the name "CSV". This data visually maps to:

```text
pokedex_number   name        type1    type2    attack   defense  speed
1                Bulbasaur   grass    poison   49       49       45
2                Ivysaur     grass    poison   62       63       60
3                Venusaur    grass    poison   82       83       80
4                Charmander  fire     -        52       43       65
5                Charmeleon  fire     -        64       58       80
6                Charizard   fire     flying   84       78       100

```

To import this dataset into a Pandas DataFrame, create a new Python program and add:

```python
import pandas as pd

df = pd.read_csv('pokemon.csv')

```

> **Note:** This code assumes the `pokemon.csv` file is in the same directory as your Python script.

If it is in a different directory, use the full path instead of the filename:

* **Windows:** `C:\Users\alex\Documents\projects\pokemon\pokemon.csv`
* **macOS:** `/Users/alex/Documents/projects/pokemon/pokemon.csv`

### Reading Directly from a URL

You don't even need to download the file first! Pandas can read directly from a web URL:

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/codedex-io/datasets/refs/heads/main/pokemon/pokemon.csv'

df = pd.read_csv(url)

```

To verify it worked, run `print(df)`. The output should look like:

```text
     no          name    type1   type2  attack  defense   speed 
0     1     Bulbasaur    grass  poison      49       49      45 
1     2       Ivysaur    grass  poison      62       63      60  
2     3      Venusaur    grass  poison      82       83      80 
3     4    Charmander     fire     NaN      52       43      65 
..  ...           ...      ...     ...     ...      ...     ...

```

### Other Separators

Not all CSV-style files use commas. TSV (Tab-Separated Values) files are fairly common, but data can theoretically use any symbol as a separator.

To load data from these file types, use `.read_csv()` with the `sep` parameter:

```python
df = pd.read_csv('pokemon.tsv', sep='\t')

```

In this case, `\t` represents a tab. If the data is separated by semicolons, use `sep=';'`.

---

### Funky Headers

Occasionally, you will find a file with unusual header formatting.

In business documents, it's fairly common for a file to begin with metadata like `Report Generated on 1/24/2026`. If imported into Pandas by default, it will incorrectly treat that first row as the column names.

To fix this, ignore the top row using the `header` parameter:

```python
df = pd.read_csv('data.csv', header=1)

```

Here, row `1` is designated as the header rather than row `0`.

#### Files Without Headers

If a file has no header at all, Pandas will assume the first row of actual data contains column names (e.g., creating a column named `Bulbasaur`).

To handle this, set `header=None` and assign custom column names using `df.columns`:

```python
df = pd.read_csv('pokemon.csv', header=None)

df.columns = ['pokedex_number', 'name', 'type1', 'type2', 'attack', 'defense', 'speed']

```

## 2. Importing Excel Files

Excel files (`.xlsx`) are widely used in business settings. They can be more complex than CSV files because a single Excel workbook can contain multiple sheets.

Import Excel files into Pandas using `pd.read_excel()`:

```python
df = pd.read_excel('sales.xlsx')

```

By default, this imports the **first sheet**.

### Specifying Sheets

To import a specific sheet, use the `sheet_name` parameter:

```python
df = pd.read_excel('sales.xlsx', sheet_name='Q1 2026')

```

To import all sheets at once, pass `sheet_name=None`. This returns a dictionary of DataFrames, where keys are sheet names and values are the corresponding DataFrames:

```python
df = pd.read_excel("sales.xlsx", sheet_name=None)

```

> **Tip:** Strategies for handling unusual headers (`header=1`, `header=None`) apply to Excel files as well!

> **Google Sheets & `.numbers`:** Pandas does not natively import Google Sheets or macOS `.numbers` files directly. Export those spreadsheets as `.csv` or `.xlsx` files first to use them with Pandas.

## 3. Importing JSON Files

JSON is the standard data format for web APIs. When pulling data from services like TikTok, Reddit, or Spotify, the response is delivered in JSON.

Here is an example JSON payload from the TikTok API:

```json
{
  "data": {
    "videos": [
      {
        "id": "7080213458555737986",
        "title": "3 Projects You Can Code in a Day",
        "video_description": "Happy coding & gaming!",
        "duration": 2,
        "create_time": 1633823999
      },
      {
        "id": "7080213458555737987",
        "title": "3 New Reads on Codedex",
        "video_description": "From new website features to project tutorials",
        "duration": 5,
        "create_time": 1633830000
      }
    ],
    "has_more": true
  },
  "error": {
    "code": "0",
    "message": "ok"
  }
}

```

### Option 1: Direct Import

You can attempt a direct import using `pd.read_json()`:

```python
df = pd.read_json('tiktok.json')

```

This creates a DataFrame where columns are named after top-level JSON keys (e.g., `data` and `error`).

### Option 2: Parsing Nested JSON

When dealing with nested structure, load the JSON file using Python's built-in `json` module first, extract the target array, and pass it directly to `pd.DataFrame()`:

```python
import pandas as pd
import json

# Load JSON file
with open('tiktok.json', 'r') as f:
    raw_json = json.load(f)

# Convert the list of videos to a DataFrame
df = pd.DataFrame(raw_json['data']['videos'])

print(df)

```

By extracting `['data']['videos']`, Pandas produces a tabular DataFrame where each row corresponds to an individual video, with columns for attributes like `id`, `title`, and `duration`.
