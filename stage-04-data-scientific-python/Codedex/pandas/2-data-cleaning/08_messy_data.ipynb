## What Exactly is "Messy Data"?

Here are four common issues you will encounter in real-world datasets:

* **Missing Data:** Think about your Codédex profile—there are many optional fields. You might add your location but leave your school blank, leaving an empty field. This is super common.
* **Duplicates:** Ever signed up for something twice by accident or clicked "Submit" too fast? Datasets frequently contain duplicate entries for the same user or item.
* **Wrong Data Types:** Python sometimes assigns incorrect data types when creating a DataFrame (e.g., treating `age` as text instead of a number), which prevents mathematical calculations.
* **Inconsistent Strings:** Text data gets messy fast. Imagine trying to filter for all users from New York when your dataset contains `'New York'`, `'new york'`, `'NYC'`, and `'New York!!!'`.

In this chapter, we will learn how to address each of these issues using Pandas!
