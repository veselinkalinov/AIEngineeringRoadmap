## Introduction

Imagine you are throwing a party and let guests RSVP via an open text box. Your responses might look like this:

```text
Name     RSVP
--------------------
Alice    Yes
Bob      no
Charlie  Y
Diana    yEs

```

If everyone answered in a standardized format, counting who said "Yes" would be a simple one-liner in Pandas. But with everyone writing differently, counting responses becomes a headache—especially in a dataset of 1,000+ people!

Luckily, Pandas allows us to clean and standardize string columns easily using vectorised string operations.

---

## Standardizing Text with `.str`

Pandas provides a `.str` accessor to apply standard Python string methods across an entire column (Series) at once.

For example, to convert all text in the `rsvp` column to lowercase:

```python
responses['rsvp'] = responses['rsvp'].str.lower()

```

### Essential `.str` Methods

| Method | Description | Example |
| --- | --- | --- |
| `.str.lower()` | Converts all characters to lowercase | `'YES'` $\rightarrow$ `'yes'` |
| `.str.upper()` | Converts all characters to uppercase | `'yes'` $\rightarrow$ `'YES'` |
| `.str.strip()` | Removes leading and trailing whitespace | `'  yes  '` $\rightarrow$ `'yes'` |
| `.str.contains('text')` | Returns `True`/`False` if a substring is present | `'Definitely'` $\rightarrow$ `True` for `'def'` |
| `.str.replace('old', 'new')` | Replaces occurrences of a pattern | `'yes'` $\rightarrow$ `'y'` |

---

## Method Chaining

You can chain multiple `.str` methods together in a single line to perform several cleaning steps at once:

```python
# Convert to lowercase and strip whitespace in one go
responses['rsvp'] = responses['rsvp'].str.lower().str.strip()

```

## Exercise: Partiful Responses

> 🥳 **The Scenario:** You've been given a DataFrame named `responses` containing 100 party RSVPs in various raw formats. Your job is to process the `rsvp` text column so every response is standardized to either `'y'` or `'n'`.

### Your Tasks

1. **Inspect the data:** Use `.head()` or `.tail()` to examine the raw RSVP responses.
2. **Standardize formatting:** Clean the `rsvp` column by lowercasing all characters and stripping out extra whitespace.
3. **Map values:** Use `.replace()` to convert `'yes'` to `'y'` and `'no'` to `'n'`.
4. **Verify counts:** Print out the final tallies using `.value_counts()`:

```python
print(responses['rsvp'].value_counts())

```

````python

import pandas as pd
import random

random.seed(1)

# Different possible RSVP variants
options = ['Yes', 'yes', 'yesss', 'Yep   ', ' yes', 'YES', 'y', 'No  ', 'no', 'N ', 'n', 'no thx']

one_hundred_responses = random.choices(options, k=100)

# Generate fake names for the 100 guests
names = [f'Guest_{i+1}' for i in range(100)]

# Create the DataFrame
responses = pd.DataFrame({
  'name': names,
  'rsvp': one_hundred_responses
})

responses.head()


````

````python
responses.tail()
````

````python
responses['rsvp'] = responses['rsvp'].str.lower().str.strip()
responses.head()
````

````python
responses['rsvp'] = responses['rsvp'].str.replace('yes', 'y').str.replace('no', 'n')
responses['rsvp'].value_counts()
````

````python
responses["rsvp"] = (
    responses["rsvp"]
    .str.replace("yss", "y")
    .str.replace("n thx", "n")
    .str.replace("yep", "y")
)

responses
````

````python
responses['rsvp'].value_counts()
````

````python
yes_response = responses[responses['rsvp'] == "y"]
yes_response
````
