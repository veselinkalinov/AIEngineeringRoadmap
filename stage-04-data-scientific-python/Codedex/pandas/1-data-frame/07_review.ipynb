Here's a recap of the main takeaways from this chapter:

    DataFrames store data in rows and columns. They can be created from dictionaries, lists, or imported directly from files.
    Some preliminary data exploration:
        .head() shows the first few rows of the DataFrame.
        .tail() shows the last few rows of the DataFrame.
        .info() displays column names, data types, etc.
        .describe() summarizes stats for numeric columns.
    Select specific columns from a DataFrame by using square brackets [ ].
    Filter rows of DataFrames by using boolean expressions using >, <, or ==. Chain multiple boolean expressions together using AND (&) and OR (|).
    We can sort, rename, and add new columns to a DataFrame.

```python
import pandas as pd

students_data = {
    "Name": [
        "Alex", "Maria", "Daniel", "Sofia", "Ivan",
        "Emma", "George", "Elena", "Martin", "Anna"
    ],
    "Age": [
        19, 20, 18, 21, 20,
        19, 22, 18, 21, 20
    ],
    "Major": [
        "Computer Science", "Engineering", "Mathematics", "Business", "Physics",
        "Computer Science", "Engineering", "Biology", "Mathematics", "Business"
    ],
    "Grade": [
        5.50, 5.20, 5.80, 4.90, 5.10,
        5.70, 4.80, 5.40, 5.60, 5.00
    ],
    "Passed": [
        True, True, True, True, True,
        True, True, True, True, True
    ]
}

students = pd.DataFrame(students_data)
students
```

```python
students['University'] = [
    "Technical University of Sofia",
    "Sofia University",
    "University of Oxford",
    "University of Cambridge",
    "MIT",
    "Stanford University",
    "RWTH Aachen University",
    "TU Delft",
    "ETH Zurich",
    "Technical University of Munich"
]
students
```

```python
grade_sorted = students.sort_values(by='Grade', ascending=True)
grade_sorted
```

```python
students = students.rename(columns = {'Name': 'First Name'})
students
```

```python
honors_student = students[students['Grade'] >= 5.5]
honors_student
```
