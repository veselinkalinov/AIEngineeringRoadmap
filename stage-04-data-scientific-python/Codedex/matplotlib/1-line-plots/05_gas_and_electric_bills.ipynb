```python
import matplotlib.pyplot as plt
import numpy as np

months = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
gas = np.array([20.32, 44.87, 67.03, 89.76, 115.34, 142.57, 167.96, 195.24, 217.88, 26.39, 23.96, 51.78])
electric = np.array([80.10, 77.79, 67.73, 64.45, 73.39, 157.13, 145.73, 91.80, 72.33, 59.77, 79.46, 93.22])

fig, axe = plt.subplots()

axe.plot(months, gas, label="Gas Bills")
axe.plot(months, electric, label="Electric Bills")

plt.title("Gas and Electric Bills")
plt.xlabel("Month")
plt.ylabel("Cost")

plt.legend()

plt.show()
```

**Bonus:** Add a third line showing the total for each month.

```python
import matplotlib.pyplot as plt
import numpy as np

months = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
gas = np.array([20.32, 44.87, 67.03, 89.76, 115.34, 142.57, 167.96, 195.24, 217.88, 26.39, 23.96, 51.78])
electric = np.array([80.10, 77.79, 67.73, 64.45, 73.39, 157.13, 145.73, 91.80, 72.33, 59.77, 79.46, 93.22])
total = gas + electric

plt.figure()
plt.plot(months, gas, label='Gas Bills')
plt.plot(months, electric, label='Electric Bills')
plt.plot(months, total, label='Total')

plt.title('Gas & Electric Bills')
plt.xlabel('Month')
plt.ylabel('Cost')

plt.legend()
plt.show()
```
