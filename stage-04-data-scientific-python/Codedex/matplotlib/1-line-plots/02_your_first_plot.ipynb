```python
import matplotlib.pyplot as plt
import numpy as np

days = np.array(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
screen_time = np.array([300, 115, 234, 317, 166, 111, 435])

plt.plot(days, screen_time)

plt.title("Screen Time of The Week")
plt.xlabel("Days of The Week")
plt.ylabel("Screen Time")

plt.show()
```
