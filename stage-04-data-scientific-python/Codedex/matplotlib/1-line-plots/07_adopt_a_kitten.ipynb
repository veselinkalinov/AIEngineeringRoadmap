Exercise link: https://www.codedex.io/matplotlib/07-adopt-a-kitten

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame(
    {
        "month": ["Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
        "adoptions": [45000, 52000, 48000, 43000, 38000, 41000],
    }
)

fig, ax = plt.subplots()

ax.plot(
    df["month"],
    df["adoptions"],
    label="Adoption Rate of Kittens",
    color="salmon",
    linestyle="-",
)

plt.title("Adoption Rate of Kittens")
plt.xlabel("Month")
plt.ylabel("Adoptions")
plt.legend()

plt.show()
```
