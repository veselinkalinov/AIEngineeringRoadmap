```python
import matplotlib.pyplot as plt
import numpy as np

games = [1, 2, 3, 4, 5]

# Points for the NBA teams
bulls_points = [89, 104, 132, 203, 115]
lakers_points = [102, 110, 95, 108, 115]
celtics_points = [98, 105, 100, 112, 107]
warriors_points = [110, 99, 105, 120, 115]

fig, ax = plt.subplots()

ax.plot(games, bulls_points, label="Chicago Bulls", color="#CE1141", linestyle="-")
ax.plot(games, celtics_points, label="Boston Celtics", color="#007A33", linestyle="-.")
ax.plot(
    games,
    warriors_points,
    label="Golden State Warriors",
    color="#1D428A",
    linestyle=":",
)
ax.plot(
    games, lakers_points, label="Los Angeles Lakers", color="#FDB927", linestyle="--"
)

plt.title("Bulls vs. Celtics vs. Warriors vs. Lakers")
plt.xlabel("Games")
plt.ylabel("Points")
plt.legend()

plt.show()
```
