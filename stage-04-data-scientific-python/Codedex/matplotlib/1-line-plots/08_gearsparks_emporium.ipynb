Exercise link: https://www.codedex.io/matplotlib/08-gearsparks-emporium

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("gearsparks_emporium_weekly_sales.csv")

fig, ax = plt.subplots()

ax.plot(
    df["Day"],
    df["Weapon"],
    label="Weapon sales per day of the week",
    color="black",
    linestyle="-",
)

plt.title("Weapon Sales at Gearspark's Emporium")
plt.xlabel("Day of the Week")
plt.ylabel("Number of Weapons Sold")
plt.legend()

plt.show()
```
