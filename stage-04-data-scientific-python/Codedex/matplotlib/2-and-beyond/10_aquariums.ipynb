Exercise link: https://www.codedex.io/matplotlib/10-aquariums

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.DataFrame({
  'aquarium': ['Monterrey Bay', 'Georgia Aquarium', 'Shedd Aquarium', 'National Aquarium'],
  'shark': [5, 4, 4, 2],
  'dolphin': [8, 12, 10, 5],
  'octopus': [7, 20, 8, 4],
  'sea turtle': [12, 5, 6, 3],
  'jellyfish': [30, 25, 20, 15]
})

fig, ax = plt.subplots(constrained_layout=True)

ax.bar(df["aquarium"], df["shark"], label="sharks per aquarium", color="b")

plt.title("Amount of sharks at each of the 4 aquariums")
plt.xlabel("Aquarium")
plt.ylabel("Sharks")
plt.legend()

plt.show()
```
