# 09. Health vs Wealth
Exercise link: https://www.codedex.io/matplotlib/09-health-vs-wealth

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('gapminder.csv')

fig, ax = plt.subplots()

ax.scatter(
    df["life_expectancy"],
    df["gdp_per_capita"],
    label="life expectancy per gdp",
    color="r",
)

plt.title("Do richer countries tend to have longer lives?")
plt.xlabel("Life Expectancy")
plt.ylabel("GDP")
plt.legend()

plt.show()
```
