```python
import matplotlib.pyplot as plt
# import numpy as np

import pandas as pd

df = pd.DataFrame({
    "Exam": ["CS", "Physics", "Calculus", "Anatomy"],
    "Alice": [90, 85, 80, 88],
    "Bob": [85, 87, 82, 90],
    "Charlie": [88, 90, 84, 86],
    "Diana": [92, 88, 85, 89]
})

fig, ax = plt.subplots(
    nrows=2,
    ncols=2,
    constrained_layout=True,
    figsize=(10, 10),
    sharex=True,
    sharey=True,
)

ax[0][0].bar(df["Exam"], df["Alice"], label="Alice", color="pink")
ax[0][0].set_title("Alice")
ax[0][0].legend()
ax[0][0].set_xlabel("Month")
ax[0][0].set_ylabel("Score")

ax[0][1].bar(df["Exam"], df["Bob"], label="Alice", color="blue")
ax[0][1].set_title("Bob")
ax[0][1].legend()
ax[0][1].set_xlabel("Month")
ax[0][1].set_ylabel("Score")

ax[1][0].bar(df["Exam"], df["Charlie"], label="Charlie", color="green")
ax[1][0].set_title("Charlie")
ax[1][0].legend()
ax[1][0].set_xlabel("Month")
ax[1][0].set_ylabel("Score")

ax[1][1].bar(df["Exam"], df["Diana"], label="Diana", color="cyan")
ax[1][1].set_title("Diana")
ax[1][1].legend()
ax[1][1].set_xlabel("Month")
ax[1][1].set_ylabel("Score")

plt.show()
```
