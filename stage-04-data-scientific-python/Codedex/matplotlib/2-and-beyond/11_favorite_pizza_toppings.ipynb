# 11. Favorite Pizza Toppings
Exercise link: https://www.codedex.io/matplotlib/11-favorite-pizza-toppings

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

toppings = np.array(
    ["Pepperoni", "Cheese", "Mushroom", "Pineapple", "Prosciutto", "Artichoke"]
)
orders = np.array([40, 25, 20, 15, 4, 5])

fig, ax = plt.subplots(constrained_layout=True)

explode = [0.1, 0, 0, 0, 0, 0]
ax.pie(orders, labels=toppings, autopct="%d%%", explode=explode)
plt.title("People's favorite pizza toppings")

plt.show()
```
