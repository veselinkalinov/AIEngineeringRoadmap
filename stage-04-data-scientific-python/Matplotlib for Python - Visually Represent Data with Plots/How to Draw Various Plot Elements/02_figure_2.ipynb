# How to plot from the pyplot module

- Using the pyplot module is the simplest method when we want to plot a single figure using (x, y). If we want to get a simple image quickly, we can call the function plot from the pyplot module, like in the example code below.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 256)
y = np.sin(x)

plt.plot(x, y)

plt.show()
```

# How to plot from the axes object

- We may remember that axes are simply a plot, so if we choose to get an axes object, we know that it can perform the same functions as a plot. We can call subplot() if we want to draw one plot. We can call subplots if we want to draw more than one plot, as the function returns multiple axes with the parameters nrows and ncols.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 256)
y = np.sin(x)

ax = plt.subplot()
ax.plot(x, y)
plt.show()
```

# How to plot from the figure object

- There are three steps to follow in order to plot from a figure object:

    - First, create a canvas (figure) explicitly.
    - Second, create a plot object by using add_subplot from the figure object.
    - Finally, plot the chart by calling plot, like in the function below.


```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 256)
y = np.sin(x)

fig = plt.figure(dpi=300)
ax = fig.add_subplot(111)
ax.plot(x, y)

plt.show()
```
