# Placing Multiple Plots onto One Figure Using Subplots

Explore how to place and arrange multiple plots within a single Matplotlib figure using the subplots function. Understand configuring rows and columns, setting figure size, and preventing element clipping with tight layout techniques to create clear and organized visual data displays.

As we have seen from previous lessons, it is possible to place more than one plot onto a single figure. In this lesson, we will learn how to place multiple plots onto one figure.

A plot is an object of the figure, meaning that we can place as many plots as we want onto a figure. Our only concern is how to arrange them.

With subplots, we are able to arrange multiple plots into a grid as long as we specify the number of rows and columns that we want, in addition to specifying the number of plots.

---

## Creating multiple subplots within one figure

`subplots()` has two options: `nrows` and `ncols`. These options return a figure instance and an array of axes objects, respectively…

The code below creates a figure with a layout containing 2 rows and 2 columns (a 2*2 arrangement).

```python
import matplotlib.pyplot as plt

fig, axe = plt.subplots(nrows=2, ncols=2)

```

````python
import matplotlib.pyplot as plt

fig, axe = plt.subplots(nrows=2, ncols=2, dpi=100)

plt.show()
````

We can easily change the layout of a figure’s subplots by changing the values of `nrows` and `ncols`. The code below draws a figure that has a layout with 2 rows and 1 column (a 2*1 arrangement).

```python
import matplotlib.pyplot as plt

fig, axe = plt.subplots(nrows=2, ncols=1)

```
---

## Setting the size of the figure

`figsize` is a parameter of `subplots()`, which is a 2-tuple float. The unit is inches.

```python
import matplotlib.pyplot as plt

fig, axe = plt.subplots(nrows=2, ncols=2, figsize=(8, 6))

```

---

## Using tight layout to avoid clipping

Matplotlib helps us automatically adjust subplots and elements so that they all fit into the figure while maintaining the logical layout requested by the user. Sometimes, our ticks, axis labels, titles, or tick labels will go outside of the figure, and will end up being clipped.

In order to prevent clipping, we can call `plt.tight_layout()`. The constrained layout helps us adjust our elements to fit them comfortably onto the figure.

````python
import matplotlib.pyplot as plt

fig, axe = plt.subplots(nrows=2, ncols=2, dpi=100)
plt.tight_layout()
axe[1][0].set_title("the third subplot")
axe[1][1].set_title("the forth subplot")

plt.show()
````
