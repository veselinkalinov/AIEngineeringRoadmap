# Setting the Title of a Plot

Explore how to set and customize titles in Matplotlib plots, including positioning, styling, and adding multiple titles to subplots and figures. Understand the difference between figure and axes titles to enhance the clarity and professionalism of your data visualizations.

The title feature allows us to set a title for our plot.

---

## Adding a title for the axes

`set_title()` is the basic function we can use to add a title to the axes. We can pass a string to this function, and a title will be added at the top center position of the graph.

```python
axe.set_title("two functions demo")

```

---

## Setting a location for the title

By default, the title is placed at the center near the top of the graph. In addition to `center`, which is the default option, we have other options: `left` and `right`. As shown in the code below, we can set the position of the title by assigning the `loc`.

```python
axe.set_title("two functions demo", loc="left")

```

> **Notice:** It is possible to add more than one title to a plot. The diagram and its accompanying code below demonstrate how to add multiple titles in different positions.

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(figsize=(7, 3.5), dpi=100)
axe.plot(points, y1)
axe.plot(points, y2)
axe.legend(["tanh", "sin"])
axe.set_title("two functions")
axe.set_title("two functions left", loc="left")
axe.set_title("two functions right", loc="right")

plt.show()
````

From line 12 to line 14, we can add three titles to a single plot in three different locations.

---

## Figure title and axes title

As we may recall from the lesson, Basics of Matplotlib, there is a hierarchical relationship between a figure and its axes. A figure can contain multiple subplots, so we can set a unique title for each of them.

As the image shows below, the red box is the figure, which is the top layer. It contains everything, includes axes, which is marked as a blue box.

Since subplots belong to a figure, they have separate attributes, including the title.

Below is an example of how to set titles for axes and figures independently. The title of the subplot is set by using `set_title()`. The title of the figure is set by using `suptitle()`.

```python
fig, axs = plt.subplots(1, 2)
axs[0].set_title("tanh function")
axs[1].set_title("sin function")
fig.suptitle("tanh & sin function")

```


````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(1, 2, figsize=(14, 3.5), dpi=300)
axe[0].plot(points, y1)
axe[0].set_title("tanh functions")

axe[1].plot(points, y2)
axe[1].set_title("sin functions")

fig.suptitle("tanh & sin function")

plt.show()
````

At line 10 and line 12, we have set the titles for the two subplots. At line 13, we have set the title for the whole figure by using `suptitle()`.

# Setting a style to a title

Similar to what we learned in the lesson, Adding Text to a Plot, we can pass a Python dict to the parameter `fontdict` to control the appearance of the title text.

Let’s compare the graph we’ve just created with the one we created in the previous lesson to see the differences between them.

```python
font = {'family': 'serif',
        'color':  'darkred',
        'weight': 'normal',
        'size': 16,
        }
axe.set_title("two functions demo", fontdict=font)

```

Finally, let’s compare the graph with the one created in lesson Grid. See what’s the difference between them.

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(nrows=1, ncols=2, dpi=800, figsize=(12, 4))

axe[0].plot(points, y1)
axe[0].plot(points, y2)
axe[0].set_xticks(np.linspace(-5, 5, 6))
axe[0].tick_params(width=2, colors="b")
axe[0].spines["right"].set_color("none")
axe[0].spines["top"].set_color("none")
axe[0].spines["left"].set_position(("data", 0))
axe[0].spines["bottom"].set_position(("axes", 0.5))
axe[0].annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round", fc="none", ec="gray"),
)
axe[0].grid(which="major", linestyle="-", linewidth="0.5", color="blue")

axe[1].plot(points, y1)
axe[1].plot(points, y2)
axe[1].set_xticks(np.linspace(-5, 5, 6))
axe[1].tick_params(width=2, colors="b")
axe[1].spines["right"].set_color("none")
axe[1].spines["top"].set_color("none")
axe[1].spines["left"].set_position(("data", 0))
axe[1].spines["bottom"].set_position(("axes", 0.5))
axe[1].annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round", fc="none", ec="gray"),
)
axe[1].grid(which="major", linestyle="-", linewidth="0.5", color="blue")
axe[1].set_title("tanh & sin function")

fig.suptitle("compare two diagram")

plt.show()
````
