# What is grid?

The grid consists of lines that connect to each of the ticks that we have set. These lines help the user locate the curve in the diagram more quickly, making the image more readable. In this lesson, we will learn how to use and how to customize the grid.

---

## How to turn on the grid

If we want to turn on the default grid, we can call `grid()`, like in the example code below. The default grid will connect to the ticks that we have set.

```python
axe.grid()

```

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(figsize=(7, 3.5), dpi=100)
axe.plot(points, y1)
axe.plot(points, y2)
axe.legend(["tanh", "sin"])
axe.annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
)
axe.grid()

plt.show()
````

At line 15, we’ve turned on the default grid by calling `grid()` without any parameters.

---

## Changing the style of the grid

The `grid()` feature allows us to modify the appearance of the grid with great detail, as we can set many options. For more information, please refer to the Grid api page on the official Matplotlib site. In this lesson, we will cover the most widely used options.

In the example code below, we set the color of the grid to blue with the line style `-`. The line width is `0.5`. `b` means blue.

```python
axe.grid(color='b', linestyle='-', linewidth=0.5)

```

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(figsize=(7, 3.5), dpi=100)
axe.plot(points, y1)
axe.plot(points, y2)
axe.legend(["tanh", "sin"])
axe.annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
)
axe.grid(color="b", linestyle="dashed", linewidth=0.5)

plt.show()
````

## Working with the major and minor ticks

By default, the grid only connects to the major ticks. However, we can still connect the grid to minor ticks. In order to do so, we must turn the minor tick on, which is set to off by default. Then, we must apply the grid to the minor tick.

In the example code below, we can see that the `which` option is used to control which ticks the grid will apply to.

```python
axe.minorticks_on()
# Customize the major grid
axe.grid(which='major', linestyle='-', linewidth='0.5', color='blue')
# Customize the minor grid
axe.grid(which='minor', linestyle=':', linewidth='0.5', color='black')

```

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(figsize=(7, 3.5), dpi=100)
axe.plot(points, y1)
axe.plot(points, y2)
axe.legend(["tanh", "sin"])
axe.annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
)
axe.minorticks_on()
axe.grid(which="major", linestyle="-", linewidth="0.5", color="blue")
axe.grid(which="minor", linestyle=":", linewidth="0.5", color="black")

plt.show()
````

Line 16 and 17 show how to turn on the grid for the major and minor ticks by setting the `which` option.

Let’s compare the graph we’ve just created with the one we made in a previous lesson, Using Annotate to Identify Data Points, to see the differences between them.

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(nrows=1, ncols=2, figsize=(14, 3.5), dpi=300)

axe[0].plot(points, y1)
axe[0].plot(points, y2)
axe[0].set_xticks(np.linspace(-5, 5, 9))
axe[0].tick_params(width=2, colors="b")
axe[0].spines["right"].set_color("none")
axe[0].spines["top"].set_color("none")
axe[0].spines["left"].set_position(("data", 0))
axe[0].spines["bottom"].set_position(("axes", 0.5))
axe[0].annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round", fc="none", ec="gray"),
)

axe[1].plot(points, y1)
axe[1].plot(points, y2)
axe[1].set_xticks(np.linspace(-5, 5, 9))
axe[1].tick_params(width=2, colors="b")
axe[1].spines["right"].set_color("none")
axe[1].spines["top"].set_color("none")
axe[1].spines["left"].set_position(("data", 0))
axe[1].spines["bottom"].set_position(("axes", 0.5))
axe[1].annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round", fc="none", ec="gray"),
)
axe[1].grid(which="major", linestyle="-", linewidth="0.5", color="blue")

plt.show()
````
