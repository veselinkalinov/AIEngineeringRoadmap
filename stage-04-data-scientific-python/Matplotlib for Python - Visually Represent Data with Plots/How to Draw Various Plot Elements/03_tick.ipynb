# What is the tick?

- The tick is a mark on the coordinate axis that shows specific points. The tick can be a number, a string, or any other notation. When plotting a graph, the axes will set a default tick based on the number ranges of the different axes. In most cases, this default setting is enough. In some cases, however, we may want to set some more customized ticks.

- Matplotlib provides many useful functions for manipulating the tick. We can set the x/y axis independently, or we can set them both with the same property.

- The following are some properties that we can manipulate by setting parameters:

    - which: Sets the x/y major/minor ticks. The default value is major.
    - color: Sets the tick color.
    - labelrotation: Rotates the tick label.
    - width: Sets the tick width in points.
    - length: Sets the tick length in points.
    - direction: Sets the direction of the ticks, either towards the axis or away from it. The default value is out.

# Setting ticks on the x-axis

- In the example below, we only set the x ticks, leaving the y ticks with their default settings. If we only want to change the tick number, set_xticks() or set_yticks() is all we need. Passing a Python list or numpy ndarray is enough

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(dpi=600)
axe.plot(points, y1)
axe.plot(points, y2)
axe.set_xticks(np.linspace(-5, 5, 9))

plt.show()
````

# Setting tick color

- In order to change the appearance of the ticks, we use tick_params().

- For example, the function below will color all major ticks blue, position them in the box, and set their dimensions to 6 points by 2 points.

```python
ax.tick_params(direction='in', length=5, width=2, colors='b')
```

Below, we can compare the left graph, which is the original graph without any settings, to the right graph, which includes what we have learned in this lesson.


````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(nrows=1, ncols=2, figsize=(12.8, 4.8), dpi=400)
axe[0].plot(points, y1)
axe[0].plot(points, y2)

axe[1].plot(points, y1)
axe[1].plot(points, y2)
axe[1].set_xticks(np.linspace(-5, 5, 6))
axe[1].tick_params(colors="b")

plt.show()
````
