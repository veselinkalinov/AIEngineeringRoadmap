# What is annotate?

- Annotate is a piece of text that identifies a specific data point on a graph, designed to help users understand charts. For example, annotate can be used to give a detailed explanation of the inflection point of a chart.

# Adding a basic annotation

- Let’s start by adding some text to one data point. The first parameter we need to define is the text that we want to display. Below are some of the other parameters that control the appearance of the annotation:

    - xy: The point (x,y)(x,y) to be annotated, which is a tuple.
    - xytext: The position (x,y) where the text will appear. None defaults to xy.
    - xycoords: The coordinate system that xy is given in.
    - textcoords: The coordinate system that xytext is given in.
    - arrowprops: The properties used to draw an arrow between xy and xytext.

-   In the example code below, the xycoords=‘data’ means that we use the coordinate system of the object being annotated. Matplotlib supports many types, which are listed below.

    | Type | Description |
    | --- | --- |
    | figure points | Points from the lower left of the figure |
    | figure pixels | Pixels from the lower left of the figure |
    | figure fraction | A fraction of the figure from the lower left |
    | axes points | Points from the lower left corner of the axes |
    | axes pixels | Pixels from the lower left corner of the axes |
    | axes fraction | A fraction of the axes from the lower left corner |
    | data | Use the coordinate system of the object being annotated |
    | polar | (theta, r), if not native ‘data’ coordinates |

```python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(dpi=100)
axe.plot(points, y1)
axe.plot(points, y2)
axe.legend(["tanh", "sin"])
axe.annotate("1.464=tanh(2)+0.5", xy=(2, 1.464), xycoords="data",
             xytext=(0.4, -40), textcoords='offset points',
             arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"))

plt.show()
```

# Changing the style of the annotation by using bbox

- The diagram below shows three separate annotations with three different style boxes. In order to create these styles, we use bbox. Although it is a time-consuming process requiring us to configure every property manually, bbox is very useful in producing high-quality pictures. Additionally, real-world data often requires us to personalize how we represent the data, so learning how to use bbox is highly recommended.

```python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(dpi=100, figsize=(9, 5))
axe.plot(points, y1)
axe.plot(points, y2)
axe.legend(["tanh", "sin"])
axe.annotate(
    "style1",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round", fc="none", ec="gray"),
)

axe.annotate(
    "style2",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(-200, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round", alpha=0.1),
)

axe.annotate(
    "style3",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(-160, -80),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round4,pad=.5", fc="0.9"),
)

plt.show()
```

As we can see from the above code, at line 15 we’ve set the border to have rounded corners by using boxstyle="round". The fc stands for fore color, which we’ve set as invisible. The ec stands for edge color, which we’ve set to gray.

At line 20, we’ve once again set the border to have rounded corners by using boxstyle="round". The alpha is set as 0.1.

Let’s compare the graph we’ve just made with the one we created in the previous lesson to see the differences between them.


```python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

# Create a figure with two axes.
fig, axe = plt.subplots(nrows=1, ncols=2, dpi=100, figsize=(12, 4))

# Plot two lines in the first axes
axe[0].plot(points, y1)
axe[0].plot(points, y2)
# Set the tick
axe[0].set_xticks(np.linspace(-5, 5, 9))
axe[0].tick_params(width=2, colors="b")
# Move the spnies like the normal cartesian coordinates
axe[0].spines["right"].set_color("none")
axe[0].spines["top"].set_color("none")
axe[0].spines["left"].set_position(("data", 0))
axe[0].spines["bottom"].set_position(("axes", 0.5))

# Plot two lines in the second axes
axe[1].plot(points, y1)
axe[1].plot(points, y2)
axe[1].set_xticks(np.linspace(-5, 5, 9))
axe[1].tick_params(width=2, colors="b")
# Move the spnies like the normal cartesian coordinates
axe[1].spines["right"].set_color("none")
axe[1].spines["top"].set_color("none")
axe[1].spines["left"].set_position(("data", 0))
axe[1].spines["bottom"].set_position(("axes", 0.5))
# Add the annotate to the second axes.
axe[1].annotate(
    "1.464=tanh(2)+0.5",
    xy=(2, 1.464),
    xycoords="data",
    xytext=(0.4, -40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.5"),
    bbox=dict(boxstyle="round", fc="none", ec="gray"),
)
plt.show()
```
