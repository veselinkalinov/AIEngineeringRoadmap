# What is the spine?

Spines are lines that connect the axis tick marks to one another. Spines also note the boundaries of the data area. They can be placed in any position we want. By default, they are the boundary of the entire axes. So, for an axes, we have four spines: top, bottom, left, and right.

In mathematics, we are used to putting the axes in the middle. In this lesson, we will learn how to move two spines into the middle, and how to make the other two invisible.

---

## Accessing the spines

In Matplotlib, there are four pre-defined spines:

* `top`
* `bottom`
* `left`
* `right`

As we can see in the example code below, spines can be used as a key to access from a `spines` attribute of the `axes` object. The key is just a string.

`ax.spines['right']`

---

## Set the spine as invisible

If we don’t want to show a spine, instead of removing it, we set its color to `none`, like in the example below.

`ax.spines['right'].set_color('none')`

In addition to providing control of the color, Matplotlib provides another function: `set_visible()`. As the name suggests, the visibility of the axes is controlled by this function.

`ax.spines['right'].set_visible(True)`

---

## Moving the spines

How to move the spine to the place you like? The spine object has the function, `set_position`, which allows us to move the corresponding spine to its desired position.

`set_position()` accepts a tuple of (position type, amount) as parameter. The position type has the following types.

* `axes`: places the spine at the specified axes coordinate (from 0.0-1.0)
* `data`: places the spine at the specified data coordinate.
* `outward`: places the spine out from the data area by the specified number of points.

In the example below, we are placing the right spine in the center.

```python
ax.spines['left'].set_position(('axes',0.5))
```

---

## We can place the spines in multiple places

In the above code, line 13 to line 16 shows how to move the four spines by using `set_position`, and how to set the colors of the spines by using `set_color`. Notice that the color of the right and the top spines are set as `none`, meaning that those spines are invisible.

Let’s compare the graph we’ve just created with the one we made in the previous lesson, Setting the Tick for a Plot, in order to see the differences between them.

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(dpi=800)
axe.plot(points, y1)
axe.plot(points, y2)
axe.set_xticks(np.linspace(-5, 5, 9))
axe.tick_params(width=2, colors="b")
axe.spines["right"].set_color("none")
axe.spines["top"].set_color("none")
axe.spines["left"].set_position(("data", 0))
axe.spines["bottom"].set_position(("axes", 0.5))

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(-5, 5, 256)
y1 = np.tanh(points) + 0.5
y2 = np.sin(points) - 0.2

fig, axe = plt.subplots(nrows=1, ncols=2, figsize=(14, 3.5), dpi=400)
axe[0].plot(points, y1)
axe[0].plot(points, y2)

axe[1].plot(points, y1)
axe[1].plot(points, y2)
axe[1].set_xticks(np.linspace(-5, 5, 7))
axe[1].tick_params(width=2, colors="b")
axe[1].spines["right"].set_color("none")
axe[1].spines["top"].set_color("none")
axe[1].spines["left"].set_position(("data", 0))
axe[1].spines["bottom"].set_position(("axes", 0.5))

plt.show()
````
