# How to Draw a Stem Plot

Explore how to draw stem plots in Matplotlib, understanding the use of vertical lines and markers to represent data points. Learn to customize stem plot styles and control baseline positions to better visualize data distribution and enhance your data representation skills.

Stem plots are similar to scatter plots because there are no links between individual data points. However, stem plots include vertical lines that connect each data point to the baseline while placing a marker at the tip. The stem plot has the advantage of being able to clearly show the shape of a data set’s distribution.

Below are some important parameters required by `stem()`.

* `x`: The position of the stems on the `x`-axis.
* `y`: The position of the stem heads on the `y`-axis.
* `linefmt`: A string defining the properties of the vertical lines.
* `markerfmt`: A string defining the properties of the markers at the stem heads.
* `bottom`: The position of the baseline on the `y`-axis.

---

## How to draw a simple stem plot

Drawing a stem plot is simple. Let’s take a look at line 8 in the example code below.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 10)
y = np.sin(x)

fig, axe = plt.subplots()

axe.stem(x, y)

axe.set_title("A Simple Stem Plot")
axe.set_xlabel("x")
axe.set_ylabel("sin(x)")
axe.grid(True)

plt.show()
```

---

## Changing the style of the line and the marker

In order to change the style of the vertical lines and the markers at the stem heads, we need to specify two parameters: `linefmt` and `markerfmt`. The first defines the style of the vertical line, while the second defines the style of the marker at the stem head. We already know about string formatting. Line 8 of the example code below demonstrates how to use it.

```python
axe.stem(x, y, linefmt='r:', markerfmt='*b')
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 10)
y = np.sin(x)

fig, axe = plt.subplots()

axe.stem(
    x,
    y,
    linefmt="r:",
    markerfmt="*b",
    basefmt="k-"
)

axe.set_title("Customized Stem Plot")
axe.set_xlabel("x")
axe.set_ylabel("sin(x)")
axe.grid(True)

plt.show()
```

---

## Changing the position of the baseline

By default, the baseline is 0, meaning that `bottom=0` and the vertical lines connect to `y=0`. `stem()` has a parameter, `bottom`, which allows us to control the baseline position.

```python
axe.stem(x, y, bottom=1.0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 10)
y = 1.5 + np.sin(x)

fig, axe = plt.subplots()

axe.stem(
    x,
    y,
    bottom=1.0
)

axe.axhline(
    1.0,
    linestyle="--",
    label="Baseline"
)

axe.set_title("Stem Plot with a Custom Baseline")
axe.set_xlabel("x")
axe.set_ylabel("y")
axe.grid(True)
axe.legend()

plt.show()
```
