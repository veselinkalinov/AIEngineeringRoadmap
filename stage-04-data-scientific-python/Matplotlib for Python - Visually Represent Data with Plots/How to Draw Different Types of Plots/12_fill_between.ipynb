# Using Fill Between to Fill the Area between Two Curves

Explore how to use Matplotlib's `fill_between()` function to fill areas between two curves. Learn to apply different colors in different regions and create confidence bands that enhance data visualization and interpretation.

Sometimes, we want to fill the areas between two curves, especially when those areas are in different quadrants, and we want to fill them with different colors. For example, when drawing a `sin(x)` curve, we may want the top and the bottom of the `x`-axis to be filled with different colors.

As another, more practical example, when we fit some data points with a line, we may want to draw a confidence region for our line, similar to how we might use an error bar plot.

In order to draw a confidence region, the basic function to call is `fill_between()`. Below are some of the important parameters required by the function:

* `x`: The `x` coordinates of the nodes defining the curves.
* `y1`: The `y` coordinates of the nodes defining the first curve.
* `y2`: The `y` coordinates of the nodes defining the second curve.
* `where`: An array of `bool` values. Defines where to exclude some horizontal regions from being filled. For example, fill between `x[i]` and `x[i + 1]` if both `where[i]` and `where[i + 1]` are `True`.

---

## How to draw a simple fill

As we can see from the example code below, `x` is the `x`-axis, whose range is `[0, 4]`. The area filled is between the `x`-axis and `y1`, which we can see at line 8.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4, 300)
y = np.sin(2 * np.pi * x)

fig, axe = plt.subplots()

axe.plot(x, y)
axe.fill_between(
    x,
    0,
    y,
    alpha=0.4
)

axe.axhline(0, linewidth=1)
axe.set_title("Area Between sin(x) and the x-axis")
axe.set_xlabel("x")
axe.set_ylabel("y")
axe.grid(True)

plt.show()
```

## How to fill a plot with different colors

As we mentioned at the beginning of this lesson, we can use different colors to fill the top and the bottom of the `x`-axis. In order to fill these areas with different colors, we can use `where`. In the example code below, since we’ve set `where=(y1 > 0)` at line 8, the area with a `y`-axis value larger than 0 is filled with green. Since we’ve set `where=(y1 < 0)` at line 9, the area with a `y`-axis value smaller than 0 is filled with red.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4, 300)
y1 = np.sin(2 * np.pi * x)

fig, axe = plt.subplots()
axe.plot(x, y1, color="black")
axe.fill_between(x, y1, where=(y1 > 0), facecolor="green", alpha=0.6, interpolate=True)
axe.fill_between(x, y1, where=(y1 < 0), facecolor="red", alpha=0.6, interpolate=True)
axe.axhline(0, linewidth=1)
axe.set_title("Different Fill Colors")
axe.set_xlabel("x")
axe.set_ylabel("y")
axe.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4, 300)
y = np.sin(2 * np.pi * x)

fig, axe = plt.subplots()

axe.plot(x, y, color="black")

axe.fill_between(
    x,
    0,
    y,
    where=(y > 0),
    facecolor="green",
    alpha=0.6,
    interpolate=True
)

axe.fill_between(
    x,
    0,
    y,
    where=(y < 0),
    facecolor="red",
    alpha=0.6,
    interpolate=True
)

axe.axhline(0, linewidth=1)
axe.set_title("Positive and Negative Areas")
axe.set_xlabel("x")
axe.set_ylabel("y")
axe.grid(True)

plt.show()
```

## How to draw a confidence band

We will be able to draw a confidence band if we follow the steps below:

* Step 1: Prepare the data that we need. In the example code below, our data is `x` and `y` at lines 4 and 5.
* Step 2: Fit our data. NumPy provides the useful function `polyfit()`. On our linear data set, we can fit a slope and an intercept, which we’ve done with `a` and `b` at line 7.

> **Notice:** `polyfit()` is a least-squares polynomial fit. It fits a polynomial `p(x) = p[0]xⁿ + p[1]xⁿ⁻¹ + ... + p[n]` to points `(x, y)`. `polyfit()` returns a vector of coefficients, `p`, that minimizes the squared error. More details can be found on the official NumPy site.

* Step 3: Get an estimation curve based on `a` and `b`, which we’ve done with `y_est` at line 8.
* Step 4: Get the error, which we’ve done with `y_err` at line 9.
* Step 5: Draw the plot by using `fill_between()`. The band is between `y_est - y_err` and `y_est + y_err`, which we’ve set at line 14.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

x = np.linspace(0, 10, 30)
y = 2.5 * x + 4 + rng.normal(0, 3, size=x.size)

slope, intercept = np.polyfit(x, y, 1)
y_est = slope * x + intercept

residuals = y - y_est
y_err = residuals.std(ddof=2)

fig, axe = plt.subplots()

axe.scatter(x, y, label="Observed data")
axe.plot(x, y_est, linewidth=2, label="Fitted line")

axe.fill_between(
    x,
    y_est - y_err,
    y_est + y_err,
    alpha=0.3,
    label="Confidence band"
)

axe.set_title("Linear Fit with a Confidence Band")
axe.set_xlabel("x")
axe.set_ylabel("y")
axe.grid(True)
axe.legend()

plt.show()
```
