# How to Draw a Box Plot

Explore how to draw box plots in Matplotlib to represent data distributions through quartiles and medians. Understand the use of key parameters to customize orientation, style, and handle outliers. This lesson guides you in preparing data sets and styling box plots for clear visualization.

---

## What is `boxplot()`?

A boxplot is a standardized way of displaying a dataset based on a five-number summary: the minimum, the maximum, the sample median, and the first and third quartiles. In statistics, the boxplot is a method for graphically depicting groups of numerical data through their quartiles.

In Matplotlib, `boxplot()` is the core function used to create a box plot. Below are some of the important parameters of the function:

* `x`: Array or a sequence of vectors. The input data.
* `vert`: Can be `True` or `False`. The default value is `True`, which displays the boxes vertically.
* `labels`: Sets the labels for each dataset.
* `notch`: Can be `True` or `False`. The default value is `False`. If `True`, the parameter will produce a notched box plot.
* `widths`: Sets the width of the box.
* `patch_artist`: Can be `True` or `False`. The default value is `False`. If `False`, the parameter will produce boxes with the `Line2D` artist. Otherwise, the boxes will be drawn with `Patch` artists.

> **Notice:** In the images below, there are some data points outside of the boxes and their whiskers. These data points are outliers. To eliminate the effect of outliers, they need to be displayed separately. Thus, the maximum or minimum value of a box plot is not necessarily the true maximum or minimum value. Instead, the values are calculated based on quartiles.
>
> * `IQR`: `Q3 - Q1`
> * `maximum`: `Q3 + 1.5 × IQR`
> * `minimum`: `Q1 - 1.5 × IQR`
>
> Data outside of these ranges are considered outliers.

## Drawing a simple box plot

In order to draw a boxplot, we need to prepare a data set. Each item in the data set is a list or tuple: one for each box. In other words, the data for the boxplot is a list of lists, a list of arrays, or a tuple of arrays.

After preparing our data set, we pass the data to `boxplot()`, like at line 13 in the code below.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

values = [
    rng.normal(loc=0, scale=1.0, size=100),
    rng.normal(loc=1, scale=1.3, size=100),
    rng.normal(loc=2, scale=0.8, size=100)
]

labels = ["Data A", "Data B", "Data C"]

fig, axe = plt.subplots()

axe.boxplot(values, labels=labels)

axe.set_title("A Simple Box Plot")
axe.set_ylabel("Value")
axe.grid(True, axis="y", alpha=0.3)

plt.show()
```

---

## Drawing a horizontal box plot

By default, the box is vertical. If we want to draw a horizontal box, we can set `vert=False`, like in the example code below.

```python
axe.boxplot(values, labels=labels, vert=False)
```

## Customizing the style of the box

If we want to customize the style of our boxplot, we can do so in many ways. We can change the color of the box, and we can customize its whiskers, caps, and the line representing the median as well. The boxplot has the following parameters:

* `boxprops`: We can specify the style of the box.
* `whiskerprops`: We can specify the style of the whisker, which is the line that connects the quartiles to the minimum or maximum.
* `medianprops`: We can specify the style of the median, which is the line that indicates the median.

They are all `dict` types.

In order to fill the box with a color, we set `patch_artist=True`, just like in the example code below.

```python
axe.boxplot(
    values,
    labels=labels,
    patch_artist=True,
    boxprops=dict(facecolor='teal', color='r')
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

values = [
    rng.normal(loc=0, scale=1.0, size=100),
    rng.normal(loc=1, scale=1.3, size=100),
    rng.normal(loc=2, scale=0.8, size=100)
]

labels = ["Data A", "Data B", "Data C"]

fig, axe = plt.subplots()

axe.boxplot(
    values,
    labels=labels,
    patch_artist=True,
    boxprops=dict(facecolor="teal", color="darkred"),
    whiskerprops=dict(color="darkred", linewidth=1.5),
    capprops=dict(color="darkred", linewidth=1.5),
    medianprops=dict(color="white", linewidth=2)
)

axe.set_title("Customized Box Plot")
axe.set_ylabel("Value")
axe.grid(True, axis="y", alpha=0.3)

plt.show()
```
