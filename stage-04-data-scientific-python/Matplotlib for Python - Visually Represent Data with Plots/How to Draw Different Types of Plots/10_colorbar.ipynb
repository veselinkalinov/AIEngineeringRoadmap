# Using Color Bars to Improve Data Readability

Explore how to use color bars in Matplotlib to improve the readability of continuous data. Understand how to create colorbars, choose colormaps, and apply normalization to map data values to colors effectively.

---

## What is the `colorbar`?

While a legend helps us identify the different elements of a chart, the feature only works for discrete data. If the data is continuous, how could we display it on the legend? A `colorbar` can help us. The `colorbar` maps the values of an interval to the colors of a range. The user can then distinguish the approximate range of the data by color. A `colorbar` is a very useful and beautiful feature.

The simplest way to draw a color bar is by calling `colorbar()`.

> **Notice:** In this lesson, `colorbar()` is called from the figure object. Let’s see some of the parameters required by `colorbar()`.
>
> * `mappable`: The `matplotlib.cm.ScalarMappable` described by the colorbar.
> * `cax`: The axes object onto which the color bar will be drawn.

---

## What is `matplotlib.cm.ScalarMappable`?

In short, `ScalarMappable` is a class used to map scalar data to RGBA. This class requires two parameters:

* `norm`: A normalization class, which typically maps a range of numbers to the interval `[0, 1]`.
* `cmap`: The colormap used to map normalized data values to RGBA colors. Matplotlib has already defined many colormaps, which we can use by calling `plt.get_cmap()` with a name as the parameter.

---

## Drawing continuous color bars

In the example code below, we are drawing a standalone color bar without other charts. In order to create such a color bar, we follow the steps below:

* Step 1: Create a figure and axes with `subplots()`.
* Step 2: Adjust the plot, which allows the label and tick to be displayed.
* Step 3: Choose a colormap by using `plt.get_cmap("viridis")`, which we’ve done at line 6.
* Step 4: Create a normalized object. In this example, the number range is `[0, 1]` at line 7.
* Step 5: Draw the `colorbar` at line 10.

> **Notice:** Line 9 shows that we need to call `set_array()`. This is a version issue. In newer versions, we don’t need to call `set_array()`.

````python
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.colors import Normalize

fig, axe = plt.subplots(figsize=(7, 1.5))

normalization = Normalize(vmin=0, vmax=1)
colormap = plt.get_cmap("viridis")

mappable = cm.ScalarMappable(
    norm=normalization,
    cmap=colormap
)
mappable.set_array([])

fig.colorbar(
    mappable,
    cax=axe,
    orientation="horizontal",
    label="Normalized value"
)

axe.set_title("Continuous Colorbar", pad=20)

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

gradient = np.linspace(0, 1, 256)
gradient = np.vstack([gradient, gradient])

colormaps = [
    "viridis",
    "plasma",
    "inferno",
    "magma",
    "cividis",
    "coolwarm"
]

fig, axes = plt.subplots(
    len(colormaps),
    1,
    figsize=(8, 5)
)

for axe, colormap in zip(axes, colormaps):
    axe.imshow(
        gradient,
        aspect="auto",
        cmap=colormap
    )
    axe.set_title(colormap, loc="left")
    axe.set_axis_off()

fig.suptitle("Comparing Colormaps")
fig.tight_layout()

plt.show()
````

## Drawing discrete color bars

Typically, the colormap is continuous. However, we are also able to draw a discrete colormap. The simplest way to do so is to pass a number to `plt.get_cmap()`, as we’ve done at line 6 in the code below. The number helps us divide the color scheme into the number of bins we want. The discrete version of a colormap is used like other colormaps.

````python
plt.get_cmap("viridis", 5)
````

````python
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.colors import BoundaryNorm

boundaries = [0, 20, 40, 60, 80, 100]
colormap = plt.get_cmap("viridis", 5)
normalization = BoundaryNorm(
    boundaries,
    colormap.N
)

fig, axe = plt.subplots(figsize=(7, 1.5))

mappable = cm.ScalarMappable(
    norm=normalization,
    cmap=colormap
)
mappable.set_array([])

fig.colorbar(
    mappable,
    cax=axe,
    boundaries=boundaries,
    ticks=boundaries,
    orientation="horizontal",
    label="Discrete value range"
)

axe.set_title("Discrete Colorbar", pad=20)

plt.show()
````
