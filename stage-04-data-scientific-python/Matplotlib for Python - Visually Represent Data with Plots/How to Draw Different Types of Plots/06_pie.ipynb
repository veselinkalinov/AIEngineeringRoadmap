# How to Draw a Pie Plot

Explore how to draw pie charts in Matplotlib by using the `pie()` function to visualize categorical data proportions. Learn to customize slices with labels, percentages via `autopct`, and emphasis with `explode`. Understand how to create nested pie charts to compare multiple datasets effectively.

---

## What is `pie()`?

Pie charts are often used to highlight the proportion of categorical data in relation to the overall value of a dataset. Pie charts are shaped like a pie, and each category that takes up a part of the pie is shaped like a slice. The larger the slice, the greater the proportion of the overall values it represents.

Now, let’s learn how to make some delicious pies.

---

## Plotting a pie chart by using `pie()`

Like with other plot types, the basic function in Matplotlib to plot a pie chart is `pie()`. Below are some of the important parameters we may need:

* `x`: The list of data we want to present for each wedge.
* `explode`: Set as `None` or as a list of values that are the same length as `x`. Each item in this list indicates how far each wedge deviates from the center of the circle.
* `labels`: A sequence of strings that provides labels for each wedge.
* `colors`: Set as `None` or as a list of colors that are the same length as `x`.
* `shadow`: Can be `True` or `False`. The default value is `False`. Setting it to `True` adds a shadow effect.
* `counterclock`: Can be `True` or `False`. The default value is `True`. Used to specify the direction of the fractions.
* `autopct`: Set as `None` or as a string to label wedges with their numeric values. The label will be placed inside the wedge.

Below is an example of how to plot a pie chart. We pass values and labels to `pie()` at line 7.

```python
import matplotlib.pyplot as plt

values = [30, 25, 20, 15, 10]
labels = ["A", "B", "C", "D", "E"]

fig, axe = plt.subplots()

axe.pie(values, labels=labels)
axe.set_title("A Simple Pie Chart")

plt.show()
```

## Add percentage information to the pie plot

The purpose of using pie charts is to see the proportions of different categories of our data, so we want to be able to show these specific proportions in the pie. `autopct` can help us because it enables us to display the percent value of a category by using Python string formatting.

In short, we can regard `autopct` as a format string, where the input is the percentage of each wedge. The string formatting when using `autopct` is consistent with the formatting that is built into Python.

In line 7 of the example code below, we can see that the formatting is `'%.2f%%'`, which indicates that two decimal places of a floating-point number will be displayed on the chart. The last two `%%` are an escape for `%`. To learn more about this topic, please refer to the official Python site.

```python
import matplotlib.pyplot as plt

values = [30, 25, 20, 15, 10]
labels = ["A", "B", "C", "D", "E"]

fig, axe = plt.subplots()

axe.pie(
    values,
    labels=labels,
    autopct="%.2f%%"
)

axe.set_title("Pie Chart with Percentages")

plt.show()
```

## Adding `explode` to the pie

Sometimes, we want to focus on a specific category, such as the one with the highest or lowest value. We can use `explode` to emphasize any number of wedges of the pie. `explode` allows us to specify an offset value for each category. As we can see in the example below, at line 5, we have set the fourth category to be offset by `0.5`.

```python
import matplotlib.pyplot as plt

values = [20, 15, 10, 30, 12, 13]
labels = ["A", "B", "C", "D", "E", "F"]
explode = [0, 0, 0, 0.5, 0, 0]

fig, axe = plt.subplots()
axe.pie(values, labels=labels, autopct="%.2f%%", explode=explode)
axe.set_title("Using explode in a Pie Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt

values = [20, 15, 10, 30, 12, 13]
labels = ["A", "B", "C", "D", "E", "F"]
explode = [0, 0, 0, 0.2, 0, 0]

fig, axe = plt.subplots()

axe.pie(
    values,
    labels=labels,
    autopct="%.2f%%",
    explode=explode,
    shadow=True
)

axe.set_title("Pie Chart with an Exploded Slice")

plt.show()
```

## Adding more than one pie to a figure

Sometimes, we may want to compare multiple data sets, meaning we need to draw two pies in one chart. In this scenario, we need to draw a nested pie chart.

As we can see from the example code below, two data sets have been drawn in one chart: one for the outer circle and the other for the inner circle.

> **Notice:**
>
> * If we want to display the labels and percentage values on the figure, we need to control the distance between them and the center of the circle.
> * `pctdistance` is used to control the distance between the center of the circle and the percentage value.
> * `labeldistance` is used to control the distance between the center of the circle and the label.

```python
import matplotlib.pyplot as plt

outer_values = [40, 30, 20, 10]
outer_labels = ["A", "B", "C", "D"]
inner_values = [25, 20, 15, 15, 10, 15]
inner_labels = ["A1", "A2", "B1", "B2", "C1", "D1"]

fig, axe = plt.subplots()
axe.pie(
    outer_values,
    radius=1.5,
    wedgeprops=dict(width=0.5),
    autopct="%.2f%%",
    pctdistance=0.8,
    labels=outer_labels,
    labeldistance=1.05,
)
axe.pie(
    inner_values,
    radius=1.0,
    wedgeprops=dict(width=0.5),
    autopct="%.2f%%",
    pctdistance=0.7,
    labels=inner_labels,
    labeldistance=0.3,
)
axe.set_title("Nested Pie Chart")
plt.show()
```

In the code below, line 11 creates an outer circle, and line 14 creates an inner circle.

```python
import matplotlib.pyplot as plt

outer_values = [40, 30, 20, 10]
outer_labels = ["A", "B", "C", "D"]

inner_values = [25, 20, 15, 15, 10, 15]
inner_labels = ["A1", "A2", "B1", "B2", "C1", "D1"]

fig, axe = plt.subplots()

axe.pie(
    outer_values,
    radius=1.3,
    labels=outer_labels,
    autopct="%.1f%%",
    pctdistance=0.82,
    labeldistance=1.08,
    wedgeprops=dict(width=0.35, edgecolor="white")
)

axe.pie(
    inner_values,
    radius=0.95,
    labels=inner_labels,
    autopct="%.1f%%",
    pctdistance=0.72,
    labeldistance=0.35,
    wedgeprops=dict(width=0.35, edgecolor="white")
)

axe.set_title("Nested Pie Chart")

plt.show()
```
