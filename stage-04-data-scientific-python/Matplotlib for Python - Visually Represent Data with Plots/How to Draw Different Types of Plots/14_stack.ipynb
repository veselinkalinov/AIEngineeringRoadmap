# How to Draw a Stack Plot

Explore how to draw stack plots using Matplotlib to visualize changing parts of a whole over time. Learn about the `stackplot()` function, its parameters, and how to adjust the baseline to enhance your data storytelling.

As we know, the pie chart is an effective way to show the proportion of individual pieces of data as they relate to the data overall. One drawback of such charts is that they only show the data at a given moment. They do not show how parts of a population change over time. We can use `stackplot()` to show how parts of the whole change over time.

`stackplot()` is the basic function that Matplotlib provides to create a stack plot. Below are some of the important parameters required by the function:

* `x`: A 1D array of dimension `N`.
* `y`: A 2D array of dimension `M × N`, or a sequence of arrays. Each array has dimension `1 × N`.

```python
stackplot(x, y)
stackplot(x, y1, y2, y3)
```

* `baseline`: The method used to calculate the baseline.
  * `'zero'`: A constant zero baseline, which is the default setting.
  * `'sym'`: Sets the plot symmetrically around zero.
  * `'wiggle'`: Minimizes the sum of the squared slopes.
* `colors`: Provides a list of colors for each data set.

## How to draw a simple stack plot

As we can see in the example code below, at line 17, we have passed a series of data in order to draw a stack plot.

```python
axe.stackplot(
    x,
    y,
    y1,
    y2,
    y3,
    y4,
    y5,
    y6,
    labels=["A", "B", "C", "D", "E", "F", "G"]
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 8)

data = [
    [2, 3, 4, 3, 5, 4, 6],
    [1, 2, 1, 3, 2, 4, 3],
    [3, 1, 2, 2, 3, 1, 2]
]

fig, axe = plt.subplots()

axe.stackplot(
    x,
    *data,
    labels=["A", "B", "C"]
)

axe.set_title("A Simple Stack Plot")
axe.set_xlabel("Time")
axe.set_ylabel("Value")
axe.legend(loc="upper left")

plt.show()
```

## Different baselines

In the example above, the baseline we used is `'zero'`, which is the default option. In the example below, we will learn about the other two options, `'sym'` and `'wiggle'`. In order to change the position of the baseline, we pass the corresponding string to the `baseline` parameter.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

data = np.vstack([
    2 + np.sin(x),
    1.5 + np.cos(x * 0.8),
    1 + 0.5 * np.sin(x * 1.5)
])

baselines = ["zero", "sym", "wiggle"]

fig, axes = plt.subplots(
    len(baselines),
    1,
    figsize=(9, 8),
    sharex=True
)

for axe, baseline in zip(axes, baselines):
    axe.stackplot(
        x,
        data,
        baseline=baseline,
        labels=["A", "B", "C"]
    )
    axe.set_title(f"baseline='{baseline}'")
    axe.set_ylabel("Value")

axes[0].legend(loc="upper right")
axes[-1].set_xlabel("Time")

fig.tight_layout()

plt.show()
```
