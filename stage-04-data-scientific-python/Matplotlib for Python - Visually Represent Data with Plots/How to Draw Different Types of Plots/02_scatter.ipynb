# How to Draw a Scatter Plot

Explore how to create scatter plots using Matplotlib's `plot()` and `scatter()` functions. Understand how to visualize relationships between two variables and represent data clusters through colors, sizes, and marker styles for clearer data insights.

A scatter plot is used to plot data points on a figure based on the horizontal and vertical axes. Scatter plots can be used to show the relationship between two variables. They can also show how data clusters in a dataset. One of the most striking features of drawing scatter plots in Matplotlib is the ability to set different colors and sizes for individual data points by using additional variables.

---

## Drawing a scatter plot from `plot()`

If we provide a series of data points, we can use `plot()` to draw them on a graph. Below is an example code that will generate data point on the `x`-axis, creating 200 data points between -4 and 4. By drawing 200 points in such a small interval, they will look like a line.

```python
import numpy as np

x = np.linspace(-4, 4, 200)
```

If we reduce the number of data points to 15 and set the `marker` at line 7 to a circle marker, we are able to see each individual data point. If we don’t set the marker, the function will draw a line that is not smooth.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

x = np.linspace(-4, 4, 15)
fig, axe = plt.subplots(dpi=300)
axe.plot(x, np.sin(x), "o")
fig.savefig("output/output.png")
plt.close(fig)
```

---

## Drawing a scatter plot from `scatter()`

Matplotlib provides the much more powerful function, `scatter()`, which allows us to control the properties of each individual data point, including size, color, edge color, and more.

We can use `scatter()` in a similar manner to `plot()`. The example code below shows us how we can call `scatter()`.

```python
axe.scatter(x, np.sin(x), 'o')
```

In addition to this basic example, below are some of the other parameters provided by `scatter`.

* `x,y`: Data points, horizontal and vertical. Commonly, these parameters are 1D arrays.
* `s`: A scalar or array, used to set the size of the marker.
* `c`: A single color string or a sequence of colors, used to set the color of the marker.
* `marker`: Sets the shape of the marker.
* `cmp`: Colormap is a way to map color, used only when `c` is an array of float.
* `alpha`: The blending value, which we can set between 0 (transparent) and 1 (opaque).

The image below is of a scatter plot that has random data points with different sizes and colors. Typically, different sizes and colors can be used to represent data in more complex ways, and can help people better understand what a chart means.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

x = np.linspace(-4, 4, 15)
fig, axe = plt.subplots(dpi=300)
axe.plot(x, np.sin(x), "o")
fig.savefig("output/output.png")
plt.close(fig)
```

At line 9, we’ve created a list of colors, and at line 10, we’ve created a list of sizes. These two lists are then passed to `scatter` as parameters.

## An example scatter plot: house price and distance

In this example, we are trying to show the relationship between the prices of some houses and their distances to some locations. In the image below, the `x`-axis represents the distance to location `x`, and the `y`-axis represents the distance to location `y`. The size of each data point indicates how high or low the price is, and the color indicates how close to the two locations the house is. By using `scatter()` and setting data points, we can easily display the relationship between house price and distance.

We can use `scatter` to show the relationship between variables, or to show the distribution of data.


````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

fig, axe = plt.subplots(dpi=100)
rng = np.random.RandomState(42)
x = rng.randn(50)
y = rng.randn(50)
colors = rng.randn(50)
sizes = rng.randn(50) * 500
axe.grid()
axe.scatter(x, y, c=colors, s=sizes, alpha=0.5)

plt.show()
````
