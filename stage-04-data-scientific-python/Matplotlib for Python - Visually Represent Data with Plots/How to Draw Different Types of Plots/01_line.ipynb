# How to Draw a Line Plot

Explore how to draw line plots in Matplotlib by plotting functions like sine waves, customizing line styles, colors, and markers, and handling multiple lines in one plot.

One of the simplest types of plots in Matplotlib is line, which draws a line defined by a function, such as `y=f(x)y=f(x)`. In this lesson, we will focus on the line.

---

## Basic line plot

Let’s begin by drawing the function `y=sin(x)y=sin(x)`. All we have to do to draw a line plot is call the `plot` function and pass the two parameters that correspond to data points on the `x` and `y` axes, respectively

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 200)
fig, axe = plt.subplots()
axe.plot(x, np.sin(x))
plt.show()
````
---

## Changing the style of the line

We can use some of the parameters listed below to draw a more complicated line. `plot()` has many parameters; the list below is not exhaustive.

* `x,y`: Data points, horizontal and vertical. Typically, these parameters are 1D arrays.
* `fmt`: A format string, e.g. ‘ro’ for red circles. We will later discuss this parameter in greater depth.
* `label`: A string, which can be used in the legend.
* `linestyle`: The style of the line, which can be ‘-’, ‘–’, ‘-.’, or ':'
* `linewidth`: The width of the line.
* `marker`: Marker style
* `color`: Marker color

The code below shows us three different ways to draw data points. We can draw data points based on an `(x,y)(x,y)` pair, we can draw only `y`, and we can draw multiple lines within the same function.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 200)
fig, axe = plt.subplots()
# plot x and y using default line style and color
axe.plot(x, np.sin(x))
# plot y using x as index array 0..N-1
axe.plot(np.sin(x))
# plot two lines at one time
axe.plot(x, np.sin(x), x+1, np.sin(x+1))
```

line 7 of the example code below shows how to draw two lines within the same function.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 200)

fig, axe = plt.subplots(dpi=300)
axe.plot(x, np.sin(x), x, np.cos(x))
fig.savefig("output/output.png")
plt.close(fig)
```

## Define line style by `fmt` string

The parameter `fmt` is a convenient way to define the basic format of a line, including color, marker, and `linestyle`. A format string consists of a part for the color, marker, and line respectively. Each one is optional, so we can still set the `linestyle`, `color`, and maker without using the `fmt`.

```python
fmt = '[marker][line][color]'
```

The table below shows the line styles supported in Matplotlib.

| Character | Description         |
| --------- | ------------------- |
| ‘-’       | solid line style    |
| ‘–’       | dashed line style   |
| ‘-.’      | dash-dot line style |
| ‘:’       | dotted line style   |

The table below shows the line colors supported by Matplotlib. These single-character color codes reflect the standard abbreviations in RGB.

| Character | Color   |
| --------- | ------- |
| ‘b’       | blue    |
| ‘g’       | green   |
| ‘r’       | red     |
| ‘c’       | cyan    |
| ‘m’       | magenta |
| ‘y’       | yellow  |
| ‘k’       | black   |
| ‘w’       | white   |

The table below shows some of the most used makers. We can check the official Matplotlib site to see the full list.

| Character | Description          |
| --------- | -------------------- |
| ‘.’       | point marker         |
| ‘o’       | circle marker        |
| ‘v’       | triangle_down marker |
| ‘^’       | triangle_up marker   |
| ‘*’       | star marker          |
| ‘s’       | square marker        |
| ‘+’       | plus marker          |

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 200)
fig, axe = plt.subplots(dpi=300)
axe.plot(x, np.sin(x), '--r' ,x, np.cos(x), '+y')
fig.savefig("output/output.png")
plt.close(fig)
```

Plotting `sin(x)` and `cos(x)` using different `fmt` strings

As we can see from the code above, the first line’s `fmt` at line 6 is `--r`, which draws a red line with dashed markers.

The second line’s `fmt` is `+y`, which draws a yellow line with plus markers.

Below is a code used to draw the lines for comparing some different line styles.

`````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 200)

fig, axe = plt.subplots(dpi=100)
axe.plot(x, x + 2, linestyle="-", color="r", marker="x", label="line1")
axe.plot(x, x + 3, linestyle="-", color="c", marker="s", label="line2")
axe.plot(x, x + 4, linestyle="-", color="m", marker="|", label="line3")
axe.plot(x, x + 5, linestyle="--", color="b", label="line4")
axe.plot(x, x + 6, linestyle="-.", color="y", label="line5")
axe.plot(x, x + 7, linestyle=":", color="b", label="line6")
axe.legend()

plt.show()
`````
