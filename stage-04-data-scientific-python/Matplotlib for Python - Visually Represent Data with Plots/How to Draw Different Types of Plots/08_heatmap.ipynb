# How to Draw a Heatmap Plot

Explore how to create heatmap plots in Matplotlib by using the `imshow()` function to visualize the relationship between two variables. Learn to add annotations for clarity and use colorbars for better interpretation, enhancing your data visualization skills.

---

## What is a heatmap?

A heatmap is a useful chart that we can use to show the relationship between two variables. A heatmap displays a general view of numerical data; it does not extract specific data points. It is a graphical representation of data where the individual values contained in a matrix are represented as colors.

In Matplotlib, there is no native function for creating a heatmap, like there is with `pie()` or `scatter()`. However, there is another function, `imshow()`, which can help us create an image with the heatmap effect.

---

## Plotting a heatmap from `imshow()`

In order to create a heatmap, we can pass a 2D array to `imshow()`. As we can see in the code below, passing the values to `imshow()` is the core operation of the plot.

After we pass the values, we can set the `x`-axis and `y`-axis labels by calling `set_xticks()` and `set_xticklabels()`.

```python
values = np.array([
    [0.8, 1.2, 0.3, 0.9, 2.2],
    [2.5, 0.1, 0.6, 1.6, 0.7],
    [1.1, 1.3, 2.8, 0.5, 1.7],
    [0.2, 1.2, 1.7, 2.2, 0.5],
    [1.4, 0.7, 0.3, 1.8, 1.0]
])

im = axe.imshow(values)
```

In the example code below, line 7 creates a 2D array from NumPy, which is considered part of the matrix. The values that we used are the correlation between two different variables. For `m_ij`, `i` is the row index and `j` is the column index. Then `m_ij` is the correlation between variables `i` and `j`.

```python
import numpy as np
import matplotlib.pyplot as plt

values = np.array([
    [0.8, 1.2, 0.3, 0.9, 2.2],
    [2.5, 0.1, 0.6, 1.6, 0.7],
    [1.1, 1.3, 2.8, 0.5, 1.7],
    [0.2, 1.2, 1.7, 2.2, 0.5],
    [1.4, 0.7, 0.3, 1.8, 1.0]
])

xlabels = ["A", "B", "C", "D", "E"]
ylabels = ["A", "B", "C", "D", "E"]

fig, axe = plt.subplots()

axe.imshow(values, cmap="viridis")

axe.set_xticks(range(len(xlabels)))
axe.set_xticklabels(xlabels)
axe.set_yticks(range(len(ylabels)))
axe.set_yticklabels(ylabels)

axe.set_title("A Simple Heatmap")
axe.set_xlabel("Variable")
axe.set_ylabel("Variable")

plt.show()
```

## Adding annotations to the heatmap

The heatmap is a striking and beautiful plot. However, if we only draw the colors, then the user will probably not understand our plot. In order to make our heatmap readable, we need to add some text onto each cell.

```python
for i in range(len(xlabels)):
    for j in range(len(ylabels)):
        text = axe.text(
            i,
            j,
            values[i, j],
            horizontalalignment="center",
            verticalalignment="center",
            color="w"
        )
```

Lines 22 and 23 add annotations to the heatmap by using `text()`. In this example, the text is the value from `values` in a double loop using indices `i` and `j`. `horizontalalignment` and `verticalalignment` center the text in each cell.

```python
import numpy as np
import matplotlib.pyplot as plt

values = np.array([
    [0.8, 1.2, 0.3, 0.9, 2.2],
    [2.5, 0.1, 0.6, 1.6, 0.7],
    [1.1, 1.3, 2.8, 0.5, 1.7],
    [0.2, 1.2, 1.7, 2.2, 0.5],
    [1.4, 0.7, 0.3, 1.8, 1.0]
])

labels = ["A", "B", "C", "D", "E"]

fig, axe = plt.subplots()

axe.imshow(values, cmap="viridis")

axe.set_xticks(range(len(labels)))
axe.set_xticklabels(labels)
axe.set_yticks(range(len(labels)))
axe.set_yticklabels(labels)

for row in range(values.shape[0]):
    for column in range(values.shape[1]):
        axe.text(
            column,
            row,
            f"{values[row, column]:.1f}",
            horizontalalignment="center",
            verticalalignment="center",
            color="white"
        )

axe.set_title("Annotated Heatmap")

plt.show()
```

## Add a color bar to the heatmap

Although we have added some text to the chart, we still want to include a special legend to show the relationship between a color and its value. A colorbar is what we need. As we can see in the example code below, all we have to do is call `axe.figure.colorbar()`, which we’ve done at line 24.

```python
im = axe.imshow(values)
axe.figure.colorbar(im, ax=axe)
```

```python
import numpy as np
import matplotlib.pyplot as plt

values = np.array([
    [0.8, 1.2, 0.3, 0.9, 2.2],
    [2.5, 0.1, 0.6, 1.6, 0.7],
    [1.1, 1.3, 2.8, 0.5, 1.7],
    [0.2, 1.2, 1.7, 2.2, 0.5],
    [1.4, 0.7, 0.3, 1.8, 1.0]
])

labels = ["A", "B", "C", "D", "E"]

fig, axe = plt.subplots()

image = axe.imshow(values, cmap="plasma")

axe.set_xticks(range(len(labels)))
axe.set_xticklabels(labels)
axe.set_yticks(range(len(labels)))
axe.set_yticklabels(labels)

axe.set_title("Heatmap with a Colorbar")

fig.colorbar(image, ax=axe, label="Value")

plt.show()
```
