```python
average_age = np.average(passengers, axis = 0)[3]
print(f"The average age of the passengers is {average_age}")
```

```python
oldest = np.max(passengers, axis = 0)[3]
print(f"The oldest passenger is {oldest} years old")

youngest = np.min(passengers, axis = 0)[3]
print(f"The youngest passenger is {youngest} years old")
```

```python
survived = np.sum(passengers, axis=0)[1]
survival_percentage = (survived / passengers.shape[0]) * 100
print(f"The survival percentage is {round(survival_percentage)}%")
```

```python
for passenger_class in [1,2,3]:
    class_passengers = passengers[passengers[:, 2]==passenger_class]

    class_survived = np.sum(class_passengers, axis=0)[1]
    class_total = class_passengers.shape[0]
    class_survival_percentage = (class_survived / class_total) * 100

    print(
        f"{class_survival_percentage}% of passengers "
        f"in class {passenger_class} survived"
    )
```
